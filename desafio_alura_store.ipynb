```python
import pandas as pd
import glob

# 1. Lista os arquivos CSV que você subiu para o Colab
arquivos = ['loja_1.csv', 'loja_2.csv', 'loja_3.csv', 'loja_4.csv']

# 2. Executa a extração e consolidação dos dados
lista_dfs = []

for arquivo in arquivos:
    try:
        df_temp = pd.read_csv(arquivo)
        lista_dfs.append(df_temp)
        print(f"Arquivo '{arquivo}' carregado com sucesso!")
    except FileNotFoundError:
        print(f"Erro: O arquivo '{arquivo}' não foi encontrado. Certifique-se de fazer o upload.")

# 3. Une todos os dados em um único DataFrame
df_final = pd.concat(lista_dfs, ignore_index=True)

# 4. Exibe o resultado da extração
print("\nExtração concluída!")
print(f"Total de registros processados: {len(df_final)}")
print("-" * 30)
print(df_final.info())
display(df_final.head())

# Opcional: Salvar o arquivo consolidado
# df_final.to_csv('dados_consolidados.csv', index=False)
```

```python

# Lista dos arquivos das lojas
arquivos = ['loja_1.csv', 'loja_2.csv', 'loja_3.csv', 'loja_4.csv']

print("--- Faturamento Total por Loja ---")

for arquivo in arquivos:
    df = pd.read_csv(arquivo)

    # Calcula a soma da coluna 'Preço'
    faturamento = df['Preço'].sum()

    # Formata o nome da loja para exibição
    nome_loja = arquivo.replace('.csv', '').replace('_', ' ').title()

    print(f"{nome_loja}: R$ {faturamento:,.2f}".replace(',', 'X').replace('.', ',').replace('X', '.'))
```

```python

# Lista dos arquivos das lojas
arquivos = ['loja_1.csv', 'loja_2.csv', 'loja_3.csv', 'loja_4.csv']

print(f"{'LOJA':<10} | {'MAIS VENDIDO (Qtd)':<30} | {'MENOS VENDIDO (Qtd)':<30}")
print("-" * 75)

for arquivo in arquivos:
    df = pd.read_csv(arquivo)
    nome_loja = arquivo.replace('.csv', '').replace('_', ' ').title()

    # Conta a frequência de cada produto
    contagem = df['Produto'].value_counts()

    # Identifica o mais e o menos vendido
    mais_vendido = contagem.idxmax()
    qtd_mais = contagem.max()

    menos_vendido = contagem.idxmin()
    qtd_menos = contagem.min()

    print(f"{nome_loja:<10} | {mais_vendido + ' (' + str(qtd_mais) + ')':<30} | {menos_vendido + ' (' + str(qtd_menos) + ')':<30}")
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Carregamento dos dados
arquivos = ['loja_1.csv', 'loja_2.csv', 'loja_3.csv', 'loja_4.csv']
df_list = []
for f in arquivos:
    temp_df = pd.read_csv(f)
    temp_df['Loja'] = f.replace('.csv', '').replace('_', ' ').title()
    df_list.append(temp_df)

df_final = pd.concat(df_list, ignore_index=True)

# 1. Gráfico de Faturamento Total por Loja
faturamento = df_final.groupby('Loja')['Preço'].sum()

plt.figure(figsize=(10, 6))
faturamento.plot(kind='bar', color='skyblue', edgecolor='black')
plt.title('Faturamento Total por Loja')
plt.xlabel('Loja')
plt.ylabel('Faturamento (R$)')
plt.xticks(rotation=45)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Carregamento dos dados
arquivos = ['loja_1.csv', 'loja_2.csv', 'loja_3.csv', 'loja_4.csv']
df_list = [pd.read_csv(f) for f in arquivos]
df_final = pd.concat(df_list, ignore_index=True)

# Conversão da coluna de data
df_final['Data da Compra'] = pd.to_datetime(df_final['Data da Compra'], dayfirst=True)

# 3. Gráfico de Evolução Mensal do Faturamento
# Agrupa por mês e soma o preço
evolucao_mensal = df_final.set_index('Data da Compra').resample('M')['Preço'].sum()

plt.figure(figsize=(12, 6))
plt.plot(evolucao_mensal.index, evolucao_mensal.values, marker='o', linestyle='-', color='green')
plt.title('Evolução Mensal do Faturamento Total (Todas as Lojas)')
plt.xlabel('Data')
plt.ylabel('Faturamento (R$)')
plt.grid(True, linestyle=':', alpha=0.6)
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Carregamento dos dados
arquivos = ['loja_1.csv', 'loja_2.csv', 'loja_3.csv', 'loja_4.csv']
df_list = []
for f in arquivos:
    temp_df = pd.read_csv(f)
    temp_df['Loja'] = f.replace('.csv', '').replace('_', ' ').title()
    df_list.append(temp_df)

df_final = pd.concat(df_list, ignore_index=True)

# 4. Gráfico de Avaliação Média por Loja (Satisfação do Cliente)
avaliacoes = df_final.groupby('Loja')['Avaliação da compra'].mean().sort_values()

plt.figure(figsize=(10, 6))
avaliacoes.plot(kind='barh', color='gold', edgecolor='black')
plt.title('Média de Avaliação dos Clientes por Loja')
plt.xlabel('Avaliação Média (1 a 5)')
plt.ylabel('Loja')
plt.xlim(0, 5)  # Garante a escala completa de 1 a 5
plt.grid(axis='x', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Carregamento dos dados
arquivos = ['loja_1.csv', 'loja_2.csv', 'loja_3.csv', 'loja_4.csv']
df_list = [pd.read_csv(f) for f in arquivos]
df_final = pd.concat(df_list, ignore_index=True)

# 6. Gráfico de Top 10 Vendedores por Faturamento
top_vendedores = df_final.groupby('Vendedor')['Preço'].sum().sort_values(ascending=False).head(10)

plt.figure(figsize=(12, 6))
top_vendedores.plot(kind='bar', color='teal', edgecolor='black')
plt.title('Top 10 Vendedores por Faturamento Total')
plt.xlabel('Vendedor')
plt.ylabel('Faturamento (R$)')
plt.xticks(rotation=45)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```
